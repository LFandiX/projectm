## Merging

Pada bagian ini kami mengelompokkan data encounter.csv dan patient.csv pada file Dataset_Hospuital_Full yang akan kami gunakan

```python
import pandas as pd 
import base64
import numpy as np

# Merging Encounters and Patients DataFrames

# Baca data
df_encounter = pd.read_csv(".\Dataset_Hospital_Full\encounters.csv")
df_patient = pd.read_csv(".\Dataset_Hospital_Full\patients.csv")

# Rename Id pada df_patient menjadi PATIENT agar bisa merge
df_patient_renamed = df_patient.rename(columns={"Id": "PATIENT"})

# Merge berdasarkan PATIENT
df_merged = pd.merge(df_encounter, df_patient_renamed, on="PATIENT", how="inner")

# Susun kolom sesuai permintaan
desired_order = [
    "Id", "PATIENT", "BIRTHDATE", "DEATHDATE", "PREFIX", "FIRST", "LAST", "SUFFIX",
    "MAIDEN", "MARITAL", "RACE", "ETHNICITY", "GENDER", "BIRTHPLACE", "ADDRESS",
    "CITY", "STATE", "COUNTY", "ZIP", "LAT", "LON", "START", "STOP", "ORGANIZATION",
    "PAYER", "ENCOUNTERCLASS", "CODE", "DESCRIPTION", "BASE_ENCOUNTER_COST",
    "TOTAL_CLAIM_COST", "PAYER_COVERAGE", "REASONCODE", "REASONDESCRIPTION"
]

# Susun dataframe dengan urutan kolom tersebut
df_final = df_merged[desired_order]

# Mabatasi jumlah baris menjadi 10000 Untuk kepentingan pengumpulan ke moodle Agar size file tidak terlalu besar
df_final = df_final.head(10000)


```

## Data Minimization

Pada bagian ini kamu menghapus data-data PII yang dinilai terlalu banyak atau inrelevan. Data tersebut adalah kolom "ORGANIZATION", 'PAYER', "PATIENT", "SUFFIX", "REASONCODE", "CODE",'BIRTHPLACE', 'STATE', 'COUNTY', 'ETHNICITY',  'LAT', 'LON', 'ZIP', 'RACE', 'DEATHDATE'.

```python
# Data Minimization 

# Menghapus data-data irrelevan yang tidak dapat diolah
kolom_kode_not_defined = ["ORGANIZATION", 'PAYER', "PATIENT", "SUFFIX", "REASONCODE", "CODE"]
data = df_final.drop(columns=[col for col in kolom_kode_not_defined if col in df_final.columns], errors='ignore')
display(data)

# Meminimalisir penganmbilan data dengan Menghapus data-data PII 
kolom_pii = ['BIRTHPLACE' ,'STATE', 'COUNTY', 'ETHNICITY',  'LAT', 'LON', 'ZIP', 'RACE', 'DEATHDATE']
data = data.drop(columns=[col for col in kolom_pii if col in data.columns], errors='ignore')
display(data)
```

## Pseudonimization 

Pada bagian ini kamu mengenkripsi data PII yang dinilai penting dan tidak boleh untuk diketahui umum taitu Kolom "FIRST","LAST","BIRTHDATE","MAIDEN","ADDRESS".

```python
# Pseudonimization 

def encode1(key, clear):
    key_len = len(key)
    key_ord = np.array([ord(k) for k in key])  # Precompute ord for key once
    clear_ord = np.array([ord(c) for c in clear])  # Precompute ord for clear text once
    enc_ord = (clear_ord + key_ord[np.tile(np.arange(key_len), len(clear)//key_len + 1)[:len(clear)]]) % 256
    return base64.urlsafe_b64encode(enc_ord.astype(np.uint8)).decode('utf-8')


#Fungsi Untuk mendekripsi data yang sudah dienkripsi
def decode1(key, enc):
    key_len = len(key)
    key_ord = np.array([ord(k) for k in key])
    enc_bytes = np.frombuffer(base64.urlsafe_b64decode(enc), dtype=np.uint8)
    clear_ord = (enc_bytes - key_ord[np.tile(np.arange(key_len), len(enc_bytes) // key_len + 1)[:len(enc_bytes)]] + 256) % 256
    return ''.join(map(chr, clear_ord))


# Melakukan enkripsi pada data-data PII yang tersisa
data["FIRST"] = data["FIRST"].apply(lambda name: encode1("secret key", str(name)))
data["BIRTHDATE"] = data["BIRTHDATE"].apply(lambda name: encode1("secret key", str(name)))
data["LAST"] = data["LAST"].apply(lambda name: encode1("secret key", str(name)))
data["MAIDEN"] = data["MAIDEN"].apply(lambda name: encode1("secret key", str(name)))
data["ADDRESS"] = data["ADDRESS"].apply(lambda name: encode1("secret key", str(name)))
display(data)
```

Kami juga melakukan penimpaan data dengan mengganti data tersebut dengan hal lain untuk mengamankan data

```python
# masking => melakukan penimpaan dat  
def masking (kolom, variabel):
    data[kolom] = variabel
    return data

# menyembunyikan identitas pasien seperti gender, penyakit dan lain-lain
data_ingin_disembunyikan = [("PREFIX", "M/F"), ("GENDER", "M/F"), ("START", "datetime"), 
                            ("STOP", "satetime"), ("REASONDESCRIPTION", "ILLNESS"), ("ENCOUNTERCLASS", 'CLASS'), 
                            ("DESCRIPTION", "DESCRIPTION")]
for column in data_ingin_disembunyikan:
    masking(column[0], column[1])

display(data)
data.to_csv("safe_data.csv", index=False)
```
